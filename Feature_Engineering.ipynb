```python
import pandas as pd
import numpy as np

print("Libraries Imported")

file_path = 'drugs.csv'
df = pd.read_csv(file_path)

df['Date'] = pd.to_datetime(df['Date'])
df.sort_values(by=['Brand_Name', 'Date'], inplace=True)

display(df.head())
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

df['Month_Sin'] = np.sin(2 * np.pi * df['Month'] / 12)
df['Month_Cos'] = np.cos(2 * np.pi * df['Month'] / 12)

print("\nTime-based features added (Year, Month, Sin, Cos):")
display(df[['Date', 'Year', 'Month', 'Month_Sin', 'Month_Cos']].head())
```

```python
df['Lag_1'] = df.groupby('Brand_Name')['Issued_Qty'].shift(1)
df['Lag_12'] = df.groupby('Brand_Name')['Issued_Qty'].shift(12)

df['Rolling_Mean_3'] = df.groupby('Brand_Name')['Issued_Qty'].transform(lambda x: x.shift(1).rolling(window=3).mean())

print("\nLag and Rolling Features added:")
display(df[['Brand_Name', 'Date', 'Issued_Qty', 'Lag_1', 'Lag_12', 'Rolling_Mean_3']].tail(15))
```

```python
df.fillna(0, inplace=True)

print("\nMissing values after lagging handled:")
print(df.isnull().sum())
```

```python
features_to_keep = [
    'Brand_Name', 'Year', 'Month', 'Month_Sin', 'Month_Cos', 
    'USD_Rate', 'Lag_1', 'Lag_12', 'Rolling_Mean_3', 'Issued_Qty'
]

df_featured = df[features_to_keep]

print("\nFinal Featured Dataset Shape:", df_featured.shape)
display(df_featured.head())
```

```python
df_featured.to_csv('drugs_featured.csv', index=False)
print("\nFeature Engineering Complete! Saved as 'drugs_featured.csv'")
```
