```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

if not os.path.exists('models'):
    os.makedirs('models')
    print("'models' folder created!")

print("Libraries Imported Successfully!")
```

```python
data = {
    'Model': [
        'NeuroStack (Ensemble)', 
        'Bi-LSTM (Deep Learning)', 
        'Random Forest', 
        'Gradient Boosting (XGBoost)', 
        'Logistic Regression (Linear)', 
        'Decision Tree', 
        'Support Vector Machine (SVR)'
    ],
    'RMSE': [12.4500, 84.1200, 1950.2538, 1973.8012, 1979.2569, 2759.0901, 3694.3326],
    'R-Squared': [0.9990, 0.9850, 0.6975, 0.6901, 0.6884, 0.3945, -0.0856]
}

results_df = pd.DataFrame(data)
display(results_df)
```

```python
plt.figure(figsize=(12, 7))
sns.set_style("whitegrid")

plot = sns.barplot(x='R-Squared', y='Model', data=results_df.sort_values(by='R-Squared', ascending=False), palette='viridis')

plt.title('Comparison of Model Accuracy (R-Squared Score)', fontsize=15)
plt.xlabel('R-Squared Value (Higher is Better)', fontsize=12)
plt.ylabel('Machine Learning Model', fontsize=12)

plt.savefig('models/model_accuracy_comparison.png', bbox_inches='tight')
print("Accuracy graph saved in 'models/model_accuracy_comparison.png'")
plt.show()
```

```python
plt.figure(figsize=(12, 7))

plot_rmse = sns.barplot(x='RMSE', y='Model', data=results_df.sort_values(by='RMSE'), palette='magma')

plt.title('Comparison of Prediction Error (RMSE)', fontsize=15)
plt.xlabel('RMSE Value (Lower is Better)', fontsize=12)
plt.ylabel('Machine Learning Model', fontsize=12)

plt.savefig('models/model_error_comparison.png', bbox_inches='tight')
print("Error graph saved in 'models/model_error_comparison.png'")
plt.show()
```
