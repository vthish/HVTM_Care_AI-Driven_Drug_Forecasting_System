```python
import joblib
import pandas as pd
from fastapi import FastAPI
from pydantic import BaseModel
from fastapi.testclient import TestClient

print("Libraries Imported Successfully")
```

```python
try:
    model = joblib.load('xgboost_base_model.pkl')
    print("Model and Label Encoder loaded successfully!")
except Exception as e:
    print(f"Error loading files. Please make sure the .pkl files exist: {e}")
```

```python
app = FastAPI(title="HVTM CARE - Predictive Model API")

class DrugInput(BaseModel):
    Year: int
    Month: int
    Month_Sin: float
    Month_Cos: float
    USD_Rate: float
    Lag_1: float
    Lag_12: float
    Rolling_Mean_3: float
    Brand_Encoded: int

@app.post("/predict")
def predict_demand(data: DrugInput):

    input_df = pd.DataFrame([data.dict()])

    prediction = model.predict(input_df)[0]

    return {
        "Status": "Success",
        "Message": "Prediction generated successfully via FastAPI",
        "Predicted_Demand_Qty": max(0, round(prediction, 2)) 
    }

print("FastAPI Service Initialized in-memory")
```

```python
app = FastAPI(title="HVTM CARE - Predictive Model API")

class DrugInput(BaseModel):
    Year: int
    Month: int
    Month_Sin: float
    Month_Cos: float
    USD_Rate: float
    Lag_1: float
    Lag_12: float
    Rolling_Mean_3: float
    Brand_Encoded: int

@app.post("/predict")
def predict_demand(data: DrugInput):

    input_df = pd.DataFrame([data.dict()])

    prediction = model.predict(input_df)[0]

    final_prediction = float(max(0, round(prediction, 2)))

    return {
        "Status": "Success",
        "Message": "Prediction generated successfully via FastAPI",
        "Predicted_Demand_Qty": final_prediction
    }

print("FastAPI Service Initialized in-memory!")
```
